## Step 1 — Confirm Python Interpreter and Version

We want to make sure you're using the **virtual environment you created** (not system Python) and a **modern 3.10+** version.


````python
import sys

print("Python executable path:", sys.executable)
print("Python version:", sys.version.split()[0])

# Quick version check
major, minor = sys.version_info[:2]
if major == 3 and minor >= 12:
    print("✅ Python version is compatible")
else:
    print("❌ Python version is not 3.10+ — please fix before continuing")
````

### Quick Install from Inside This Notebook

If you’d rather have the notebook do it for you (same effect as above), run the cell below.

````python
import sys
from pathlib import Path
import subprocess

req_file = Path("requirements.txt")
if not req_file.exists():
    print("❌ No requirements.txt found in this folder. Please add it and rerun.")
else:
    print("📦 Installing packages from requirements.txt (this may take a while)...")
    subprocess.check_call([sys.executable, "-m", "pip", "install", "-r", str(req_file)])
    print("✅ Packages installed successfully")
````
